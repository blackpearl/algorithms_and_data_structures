```python
import hashlib

def hash_to_range(input_string: str) -> int:
     hash_object = hashlib.sha256(input_string.encode())
     hash_int = int(hash_object.hexdigest(), 16)
     return (hash_int % 3) + 1
input_string = "Rajthilak"
result = hash_to_range(input_string)
print(result)
```

  # Question Two: Path to Leaves

  Given the `root` of a binary tree, return all root to leaf paths in any order.

  ## Examples

  ### Example 1

  ![](./images/q1_ex1.png)

  Input: `root = [1, 2, 2, 3, 5, 6, 7]` *What traversal method is this?*

  Output: [[1, 2, 3], [1, 2, 5], [1, 2, 6], [1, 2, 7]]

  ### Example 2

  ![](./images/q1_ex3.png)

  Input: `root = [10, 9, 7, 8]`

  Output: [[10, 7], [10, 9, 8]]

</details>

#### Starter Code for Question 2

```python
from typing import List, Optional

# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right
def bt_path(root: TreeNode) -> List[List[int]]:
  def dfs(node, path):
        if not node:
            return
        path.append(node.val)
        if not node.left and not node.right:  # If it's a leaf node
            paths.append(path[:])  # Add a copy of the path
        else:
            dfs(node.left, path)
            dfs(node.right, path)
        path.pop()  # Backtrack

  paths = []
  dfs(root, [])
  return paths

# Helper function to create a binary tree from a list (level-order input)
def build_tree(values: List[Optional[int]]) -> Optional[TreeNode]:
    if not values:
        return None
    nodes = [TreeNode(val) if val is not None else None for val in values]
    kids = nodes[::-1]
    root = kids.pop()
    for node in nodes:
        if node:
            if kids:
                node.left = kids.pop()
            if kids:
                node.right = kids.pop()
    return root

# Example usage
example1 = build_tree([1, 2, 2, 3, 5, 6, 7])
example2 = build_tree([10, 9, 7, 8])

print("Example 1 Output:", bt_path(example1))
print("Example 2 Output:", bt_path(example2))
```

-   Paraphrase the problem in your own words.


The give input should be kept in original order but all the zeroes should be moved to the end. But the non-zero should be kept in the same order as it is in the input.


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


Input: nums = [0, 8, 0, 3, 9]

Output: [8, 3, 9, 0, 0]

The give input non-numeric numbers 8,3,9 are put in the same order as it is in the input, and whatever number of zeroes left are appended to the end.


-   Copy the solution your partner wrote. 


```python
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    for i,num1 in enumerate(nums):
        if num1==0:
            for j,num2 in enumerate(nums[i+1:]):
                if num2!=0:
                    nums[i],nums[j+i+1]=nums[j+i+1],nums[i]
                    break
    return nums

move_zeros_to_end([0, 1, 0, 3, 12])
```


-   Explain why their solution works in your own words.


The function goes through the list one by one. If the number is 0, it finds the next number that is not 0, after the current position, and swaps the non-zero number to zero number position. By doing this, non-zero numbers are moved forward, and zeros move toward the end.


-   Explain the problem’s time and space complexity in your own words.


The outer loop goes through each element of the list once so the time complexity O(n). However there is an inner loop, and, in the worst case, for each zero, this can scan the whole remaining list, so another O(n). The resulting time complexity is almost O(n2) (n square). The space complexity is O(1), as it works on the same list without any temps or duplicate lists.


-   Critique your partner's solution, including explanation, and if there is anything that should be adjusted.


I think the only one disadvantage in the given solution could be the time complexity n-square. Having two loops, always increase the time complexity.  This should be corrected, the solution should be given in one loop instead of loop within a loop.
